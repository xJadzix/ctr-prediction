Import Polars

```python
import polars as pl
from pathlib import Path

# Path to the raw training data
DATA_PATH = Path("../data/raw/train.gz")

# Sanity check
print(f"File exists: {DATA_PATH.exists()}")
print(f"File size: {DATA_PATH.stat().st_size / 1e9:.2f} GB")
```

### Data structure

```python
# Scan the file without reading data - only reads schema
df_lazy = pl.scan_csv(DATA_PATH) # "lazy" API

# Get schema (column names and types)
schema = df_lazy.collect_schema()

print(f"Number of columns: {len(schema)}")
print("\nSchema:")
for col_name, dtype in schema.items():
    print(f" {col_name}: {dtype}")
```

```python
# Load a sample of the data (first 10 rows for inspection)
df_head = pl.read_csv(DATA_PATH, 
                      n_rows=10,
                     schema_overrides={"id": pl.String}) # override 'id' type to String (values are too big for Int64)

# Display the sample
df_head
```
